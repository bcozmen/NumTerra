```python
from numba import njit, prange
import numpy as np
import matplotlib.pyplot as plt
from tqdm import trange

@njit(cache=True)
def diamond_square_numba(grid, scale, roughness, seed=0):
    size = grid.shape[0]

    # deterministic LCG RNG
    state = np.uint64(seed if seed is not None else 0)

    def next_rand(s):
        s = np.uint64((s * np.uint64(6364136223846793005) +
                       np.uint64(1442695040888963407)) &
                      np.uint64(0xFFFFFFFFFFFFFFFF))
        return s, (s >> np.uint64(11)) / float(np.uint64(1) << np.uint64(53))

    def is_fixed(x):
        return not np.isnan(x)

    # -------------------------
    # initialize corners only if not fixed
    # -------------------------
    state, r = next_rand(state)
    if not is_fixed(grid[0, 0]):
        grid[0, 0] = r * scale

    state, r = next_rand(state)
    if not is_fixed(grid[0, size - 1]):
        grid[0, size - 1] = r * scale

    state, r = next_rand(state)
    if not is_fixed(grid[size - 1, 0]):
        grid[size - 1, 0] = r * scale

    state, r = next_rand(state)
    if not is_fixed(grid[size - 1, size - 1]):
        grid[size - 1, size - 1] = r * scale

    step = size - 1
    current = scale

    while step > 1:
        half = step // 2

        # =========================
        # Square step
        # =========================
        for x in range(half, size - 1, step):
            for y in range(half, size - 1, step):

                if is_fixed(grid[x, y]):
                    continue

                avg = (
                    grid[x - half, y - half] +
                    grid[x - half, y + half] +
                    grid[x + half, y - half] +
                    grid[x + half, y + half]
                ) * 0.25

                state, r = next_rand(state)
                grid[x, y] = avg + (r - 0.5) * current

        # =========================
        # Diamond step
        # =========================
        for x in range(0, size, half):

            start_y = half if ((x // half) % 2 == 0) else 0

            for y in range(start_y, size, step):

                if is_fixed(grid[x, y]):
                    continue

                total = 0.0
                count = 0

                if x - half >= 0:
                    total += grid[x - half, y]
                    count += 1

                if x + half < size:
                    total += grid[x + half, y]
                    count += 1

                if y - half >= 0:
                    total += grid[x, y - half]
                    count += 1

                if y + half < size:
                    total += grid[x, y + half]
                    count += 1

                state, r = next_rand(state)
                grid[x, y] = total / count + (r - 0.5) * current

        step //= 2
        current *= roughness

    return grid
```

```python
def bilinear(grid, x, y):
    h, w = grid.shape

    x0 = int(np.clip(x, 0, h - 2))
    y0 = int(np.clip(y, 0, w - 2))

    x1 = x0 + 1
    y1 = y0 + 1

    tx = x - x0
    ty = y - y0

    v00 = grid[x0, y0]
    v10 = grid[x1, y0]
    v01 = grid[x0, y1]
    v11 = grid[x1, y1]

    return (
        v00 * (1 - tx) * (1 - ty) +
        v10 * tx * (1 - ty) +
        v01 * (1 - tx) * ty +
        v11 * tx * ty
    )

def make_warp_field(size, scale=80, seed=999):
    np.random.seed(seed)

    field = np.random.randn(size, size).astype(np.float32)

    # cheap smoothing (box blur iterations)
    for _ in range(5):
        field = (
            np.roll(field, 1, 0) +
            np.roll(field, -1, 0) +
            np.roll(field, 1, 1) +
            np.roll(field, -1, 1) +
            field
        ) / 5.0

    # normalize
    field -= field.min()
    field /= (field.max() + 1e-8)

    return field * scale

def warp_sample(heightmap, wx, wy, strength=25):
    h, w = heightmap.shape
    out = np.zeros_like(heightmap)

    for x in range(h):
        for y in range(w):

            sx = x + wx[x, y] * strength
            sy = y + wy[x, y] * strength

            sx = np.clip(sx, 0, h - 2)
            sy = np.clip(sy, 0, w - 2)

            out[x, y] = bilinear(heightmap, sx, sy)

    return out
```

```python
size = 2**8 + 1
roughness = 0.6

tiles = 4

grids = [[None for _ in range(tiles)] for _ in range(tiles)]
seeds = 55

for i in range(tiles):
    for j in range(tiles):

        grid = np.full((size, size), np.nan, dtype=np.float32)

        if i > 0:
            grid[0, :] = grids[i - 1][j][-1, :]

        if j > 0:
            grid[:, 0] = grids[i][j - 1][:, -1]

        grids[i][j] = diamond_square_numba(
            grid=grid,
            scale=1,
            roughness=roughness,
            seed=seeds
        )

        seeds += 1
# -----------------------------
# stitch full terrain
# -----------------------------
big = np.block(grids)

# -----------------------------
# plot stitched terrain
# -----------------------------
plt.figure(figsize=(10, 10))
plt.imshow(big, cmap="terrain")
plt.title("2x2 Diamond-Square Tiled Terrain")
plt.axis("off")
plt.show()
```
